```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

#load dataset Iris dari scikit-Learn 
iris = load_iris()
```

pada cell ini mempersiapkan library dan dataset yang akan digunakan. numpy untuk mengolah data dalam bentuk array, sedangkan pandas digunakan untuk mengolah data dalam bentuk tabel. load_iris digunakan untuk mengambil dataset Iris yang sudah tersedia di Scikit-learn. Setelah itu, iris = load_iris() menyimpan dataset tersebut ke dalam variabel iris.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

pd.DataFrame() mengubah dataset menjadi DataFrame. X berisi 4 fitur bunga, yaitu panjang dan lebar sepal serta panjang dan lebar petal, sedangkan y berisi jenis bunga yang akan diprediksi.

```python
y
```

digunakan untuk menampilkan isi dari variabel y. Variabel tersebut berisi jenis bunga Iris, yaitu setosa, versicolor, dan virginica. Input ini hanya digunakan untuk melihat data target sebelum dilakukan proses selanjutnya.

```python
from sklearn.preprocessing import LabelEncoder

lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

LabelEncode digunakan untuk mengubah nama bunga menjadi angka lab = LabelEncoder() membuat objek untuk melakukan perubahan tersebut, kemudian lab.fit_transform(y) mempelajari label yang tersedia dan mengubahnya menjadi angka.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

train_test_split(). test_size=0.2digunakan untuk membagi dataset menjadi data training dan testing berarti 20% data digunakan sebagai data testing dan 80% sebagai data training. random_state=0 digunakan agar pembagian data tetap sama ketika program dijalankan kembali. X_train dan y_train digunakan untuk melatih model, sedangkan X_test dan y_test digunakan untuk menguji model.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

GaussianNB() digunakan untuk membuat model Gaussian Naive Bayes, Model ini kemudian dilatih menggunakan fit(X_train, y_train), yaitu dengan mempelajari hubungan antara ciri-ciri bunga pada X_train dengan jenis bunga pada y_train. Setelah proses tersebut, model yang sudah dilatih disimpan dalam variabel NB_model.

```python
result = NB_model.predict(X_test)
result
```

melakukan prediksi terhadap data testing. NB_model.predict(X_test) meminta model untuk menentukan jenis bunga berdasarkan data X_test. Hasil prediksi disimpan dalam variabel result dan masih berupa angka

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

accuracy_score() digunakan untuk menghitung tingkat ketepatan model Fungsi ini membandingkan y_test sebagai jawaban sebenarnya dengan result sebagai hasil prediksi model. Hasilnya kemudian dikalikan 100 agar ditampilkan dalam bentuk persentase

```python
#Prediksi inputan data baru
# input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
